```python
import re
```

```python
with open('/home/marco/Rosalind/6idmapping_2025_11_23.fasta') as f:
    DNA = f.read().split('\n')
```

```python
print(type(DNA))
DNA
```

```python
informações = []
indices_cabeçalhos = []
for item in DNA:
    if item.startswith('>sp'):
        informações.append(item)
        indices_cabeçalhos.append(DNA.index(item))
print(informações)
indices_cabeçalhos
```

```python
valores_id = []
for item in informações:
    item = item.lstrip('>sp').split(' ')[0]
    item = item.split('|')
    item = '_'.join(item[1:])
    valores_id.append(item)
```

```python
valores_id
```

```python
sequencias = []
for i in range(len(indices_cabeçalhos)-1):
    inicio_seq = indices_cabeçalhos[i] + 1
    final_seq = indices_cabeçalhos[i+1]
    sequencias.append(''.join(DNA[inicio_seq:final_seq]))

sequencias.append(''.join(DNA[final_seq + 1:]))
```

```python
sequencias
```

```python
id_sequencia = dict(zip(valores_id,sequencias))
id_sequencia
```

```python
motivo = r"(?=(N[^P][ST][^P]))"
results_dict = {}
```

```python
for valores_id, sequencia in id_sequencia.items():
    indices_sequencia = []
    for correspondencia in re.finditer(motivo, sequencia):
        inicio_motivo = correspondencia.start() + 1
        indices_sequencia.append(inicio_motivo)
    results_dict[valores_id] = indices_sequencia
```

```python
results_dict
```

```python
for valores_id, indices in results_dict.items():
    if indices:
        print(f"\n{valores_id}")
        for i in range(len(indices)):
            print(f"{indices[i]}", end=' ') 
print()
```
