```python
import requests
import re
```

```python
with open('/home/marco/Rosalind/rosalind_mprt.txt') as f:
    ids = f.read().split('\n')
    ids.pop()
```

```python
#Retirar somente o ID que é utilizado para fazer a pesquisa no uniprot.
uniprot_ids = []

for id in ids:
    id = id.split('_')
    uniprot_ids.append(id[0])
    #print(id)
```

```python
print(uniprot_ids)
```

```python

# Extratir as sequências fasta através de um pedido (request) no site da uniprot. 
sequencias_fasta = []

for uniprotId in uniprot_ids:
    
    url = f"http://www.uniprot.org/uniprot/{uniprotId}.fasta"
    
    try:
        # Faz a requisição GET
        response = requests.get(url)
        response.raise_for_status()

        # Obtém o conteúdo da sequência
        fasta_content = response.text
        print(fasta_content)
        %timeit

        # Processamento do texto: Separa o cabeçalho e as quebras de páginas ('\n')
        linhas = fasta_content.split('\n')

        # Une as partes da sequência em uma única sequência e armazena cada sequência em uma lista  
        sequencia = "".join(linha.strip() for linha in linhas[1:] if linha.strip())
        sequencias_fasta.append(sequencia)

    
    except requests.exceptions.RequestException as e:
        print(f"Erro ao acessar {url}: {e}")
        
```

```python
sequencias_fasta
```

```python
dict_ids_sequencias = dict(zip(ids, sequencias_fasta))
dict_ids_sequencias
```

```python
# r = raw; ?= Inclue padrões que se sobrepõem
# [^P] = qualquer aminoácido exceto Prolina (P)
# [ST] = Serina (S) OU Treonina (T)
motivo = r"(?=(N[^P][ST][^P]))"
```

```python
results_dict = {}

for ids, sequencia in dict_ids_sequencias.items():
    indices_sequencia = []
    for correspondencia in re.finditer(motivo, sequencia):
        inicio_motivo = correspondencia.start() + 1
        indices_sequencia.append(inicio_motivo)
    results_dict[ids] = indices_sequencia
```

```python
results_dict
```

```python
for valores_id, indices in results_dict.items():
    if indices:
        print(f"\n{valores_id}")
        for i in range(len(indices)):
            print(f"{indices[i]}", end=' ') 
print()
```
