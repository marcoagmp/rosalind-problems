```python
with open('/home/marco/Rosalind/rosalind_orf.txt') as f:
    arquivo = f.read().split('\n')
    arquivo.pop()
    sequencia = "".join(arquivo[1:])
```

```python
sequencia
```

```python
rna_m = []

for i in range(0,len(sequencia),3):
    codon = ''.join(sequencia[i:i+3])
    rna_m.append(codon)
```

```python
#rna_m.pop()
print(rna_m)
```

```python
proteina = []
for codon in rna_m:
    if codon in aminoacidos.keys():
        proteina.append(aminoacidos[codon])
#print(proteina)
proteina = ''.join(proteina[:])
print(type(proteina))
print(proteina)
```

```python
metionina = re.finditer('M', proteina)
ORFs = []

for M in metionina:
    print(M.start())
    start_index = M.start()
    sub_sequencia = proteina[start_index:]
    #print(sub_sequencia)

    
    stop_index = sub_sequencia.index('*')
    ORF = sub_sequencia[:stop_index]
    print()
    print(ORF)
    print()
    ORFs.append(ORF)
```
