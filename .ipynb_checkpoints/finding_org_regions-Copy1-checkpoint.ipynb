```python
import re
```

```python
aminoacidos = {
    'TTT': 'F', 'TTC': 'F', 'TTA': 'L', 'TTG': 'L',
    'TCT': 'S', 'TCC': 'S', 'TCA': 'S', 'TCG': 'S',
    'TAT': 'Y', 'TAC': 'Y', 'TAA': '*', 'TAG': '*',
    'TGT': 'C', 'TGC': 'C', 'TGA': '*', 'TGG': 'W',

    'CTT': 'L', 'CTC': 'L', 'CTA': 'L', 'CTG': 'L',
    'CCT': 'P', 'CCC': 'P', 'CCA': 'P', 'CCG': 'P',
    'CAT': 'H', 'CAC': 'H', 'CAA': 'Q', 'CAG': 'Q',
    'CGT': 'R', 'CGC': 'R', 'CGA': 'R', 'CGG': 'R',

    'ATT': 'I', 'ATC': 'I', 'ATA': 'I', 'ATG': 'M',
    'ACT': 'T', 'ACC': 'T', 'ACA': 'T', 'ACG': 'T',
    'AAT': 'N', 'AAC': 'N', 'AAA': 'K', 'AAG': 'K',
    'AGT': 'S', 'AGC': 'S', 'AGA': 'R', 'AGG': 'R',

    'GTT': 'V', 'GTC': 'V', 'GTA': 'V', 'GTG': 'V',
    'GCT': 'A', 'GCC': 'A', 'GCA': 'A', 'GCG': 'A',
    'GAT': 'D', 'GAC': 'D', 'GAA': 'E', 'GAG': 'E',
    'GGT': 'G', 'GGC': 'G', 'GGA': 'G', 'GGG': 'G'
}
aminoacidos.keys()
```

```python
sequencia = 'AGCCATGTAGCTAACTCAGGTTACATGGGGATGACCCCGCGACTTGGATTAGAGTCTCTTTTGGAATAAGCCTGAATGATCCGAGTAGCATCTCAG'

bases = {'A':'T', 'G':'C', 'C':'G', 'T':'A'}
sequencia_inversa = sequencia[::-1]

print(sequencia)
#print()
#print(sequencia_inversa)
#print(type(sequencia_inversa))

sequencia_reversa = []
for base in sequencia_inversa:
    base = bases[base]
    sequencia_reversa.append(base)

sequencia_reversa = ''.join(sequencia_reversa[:])
#print()
print(sequencia_reversa)
```

```python
def transcrição(seq):
    rna_m = []

    for i in range(0,len(seq),3):
        codon = ''.join(seq[i:i+3])
        rna_m.append(codon)

    return rna_m
```

```python
rna_m = transcrição(sequencia)
rna_m_reverso = transcrição(sequencia_reversa)
print(rna_m)
print()
print(rna_m_reverso)
```

```python
def tradução(rna_m):
    
    proteina = []
    for codon in rna_m:
        if codon in aminoacidos.keys():
            proteina.append(aminoacidos[codon])
    #print(proteina)
    proteina = ''.join(proteina[:])
    #print(type(proteina))
    #print(proteina)

    return proteina
```

```python
proteina_fwd = tradução(rna_m)
proteina_rev = tradução(rna_m_reverso)
print(proteina_fwd)
print()
print(proteina_rev)
```

```python
def extrai_ORF(proteina):
    
    # Para garantir que o código funcione, re-executamos a busca por 'M'
    metionina = re.finditer('M', proteina)
    ORFs = []
    
    print("--- Buscando e Extraindo ORFs ---")
    
    for M in metionina:
        start_index = M.start()
        sub_sequencia = proteina[start_index:]
        
        print(f"\nInício 'M' no índice: {start_index}")
        print(f"Sub-sequência: {sub_sequencia}")
    
        # 1. Usar .find('*') para obter o índice. Retorna -1 se não encontrar.
        stop_relative_index = sub_sequencia.find('*')
        
        # 2. Verificar se o códon de parada foi encontrado (índice >= 0)
        if stop_relative_index != -1:
            
            # O códon de parada foi encontrado: extrai a ORF antes dele
            ORF = sub_sequencia[:stop_relative_index]
            
            print(f"Códon de parada encontrado no índice relativo: {stop_relative_index}")
            print(f"ORF: {ORF}")
            ORFs.append(ORF)
            
        else:
            # O códon de parada NÃO foi encontrado: a ORF é a sub_sequencia inteira
            ORF = sub_sequencia
            
            print("Aviso: Códon de parada (*) não encontrado.")
            print(f"ORF: {ORF}")
            ORFs.append(ORF)
    
    print("\n--- Lista Final de ORFs ---")
    print(ORFs)
    return ORFs
```

```python
ORFs_fwd = extrai_ORF(proteina_fwd)
ORFs_rev = extrai_ORF(proteina_rev)
```
